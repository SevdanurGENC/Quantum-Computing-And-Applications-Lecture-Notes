## Kuantum Hesaplamaya Giriş: Temel Kavramlar

### 1\. Kuantum Mekaniği: Neden Farklı?

Geleneksel bilgisayarlarımız, **bit** adı verilen birimlerle çalışır. Bir bit ya **0** ya da **1** değerini alabilir.

Kuantum hesaplama ise, maddenin çok küçük ölçekteki (atom altı parçacıklar) davranışlarını yöneten fizik teorisi olan **Kuantum Mekaniği** prensiplerini kullanır. Bu prensipler, geleneksel bilgisayarların yapamayacağı bazı şeyleri yapmamızı sağlar:

  * **Süperpozisyon (Superposition):** Bir parçacığın aynı anda birden fazla durumda bulunabilmesi.
  * **Dolanıklık (Entanglement):** İki veya daha fazla parçacığın, aralarındaki mesafe ne olursa olsun, birbiriyle anlık olarak ilişkilendirilmesi.

### 2\. Qubit (Kuantum Biti)

Geleneksel bitten farklı olarak, kuantum bilgisayarların temel bilgi birimi **Qubit**'tir (Kuantum Biti).

  * **Klasik Bit:** Ya **0**'dır ya da **1**.
  * **Qubit:** Aynı anda hem **0** hem de **1** olabilir. Sadece ölçtüğünüz anda tek bir değere (0 veya 1) çöker.

#### Qubit Durumları ve Braket Gösterimi

Qubit'in matematiksel durumunu ifade etmek için **Braket Gösterimi** (Dirac Gösterimi) kullanılır.

  * **Ket ($|\psi\rangle$):** Bir kuantum durum vektörünü temsil eder.

| Qubit Durumu | Anlamı | Vektör Gösterimi | Braket Gösterimi |
| :--- | :--- | :--- | :--- |
| **Temel Durum** | Kesinlikle 0 | $\begin{pmatrix} 1 \\ 0 \end{pmatrix}$ | $|0\rangle$ (Ket zero) |
| **Temel Durum** | Kesinlikle 1 | $\begin{pmatrix} 0 \\ 1 \end{pmatrix}$ | $|1\rangle$ (Ket one) |
| **Süperpozisyon** | Genel Durum | $\alpha |0\rangle + \beta |1\rangle$ | $|\psi\rangle$ |

**Not:** Burada $\alpha$ ve $\beta$, **olasılık genlikleridir**. Bu genliklerin karelerinin toplamı daima 1'dir: $|\alpha|^2 + |\beta|^2 = 1$. ($|\alpha|^2$, ölçüm sonucunun 0 olma olasılığını verir.)

### 3\. Süperpozisyon: Hem Var Hem Yok

Süperpozisyon, bir qubit'in aynı anda birden fazla temel durumu ($|0\rangle$ ve $|1\rangle$) bir kombinasyonu olarak bulunabilmesidir.

**Örnek (Klasik):** Bir lamba ya **açıktır** ya da **kapalıdır**.
**Örnek (Kuantum):** Bir qubit, ölçülene kadar belirli bir olasılıkla hem **açık** hem de **kapalı** durumundadır.

**Cirq Kod Örneği: $|0\rangle$ durumundaki Qubit'i Süperpozisyona Getirme**

Süperpozisyon yaratmak için en sık kullanılan kapı **Hadamard Kapısı**'dır ($H$).

```
!pip install cirq
```

```
import cirq

# 1. Qubit'i tanımlayalım
qubit = cirq.GridQubit(0, 0)

# 2. Kuantum Devresini oluşturalım
circuit = cirq.Circuit(
    # Başlangıçta |0> durumundadır.
    # Hadamard (H) kapısını uygulayarak süperpozisyon oluşturuyoruz:
    cirq.H(qubit),
    # Durumu görmek için ölçüm yapalım
    cirq.measure(qubit, key='result')
)

# 3. Devreyi simüle edelim ve 1000 kez çalıştıralım
simulator = cirq.Simulator()
results = simulator.run(circuit, repetitions=1000)

print("Kuantum Devresi:\n", circuit)
print("\nÖlçüm Sonuçları (1000 Çalıştırma):")
print(results.histogram(key='result'))
```

**Çıktı Yorumu:**
$H|0\rangle$ uygulandığında qubit, $\frac{1}{\sqrt{2}}|0\rangle + \frac{1}{\sqrt{2}}|1\rangle$ durumuna gelir. Yani 0 ve 1 olma olasılığı yaklaşık olarak %50'dir. Çıktıda 0 ve 1 sayılarının birbirine yakın olduğunu göreceksiniz (yaklaşık 500'e 500).

### 4\. Dolanıklık (Entanglement): Uzaktan Bağlantı

Dolanıklık, iki veya daha fazla qubit'in durumlarının birbiriyle **anlık olarak** ilişkilendirilmesidir. Bir qubit'i ölçtüğünüz anda, aralarındaki mesafe ne olursa olsun, diğer dolanık qubit'in durumunu da **hemen** bilirsiniz.

**Örnek:** İki madeni para dolanıktır. Birini attınız ve **Tura** geldiğini gördünüz. Eğer dolanık iseler, diğer parayı atmasanız bile **Yazı** geleceğini anında bilirsiniz (veya ikisi de aynı şekilde dolanık ise, o da Tura gelecektir).

**Cirq Kod Örneği: İki Qubit'i Dolanık Hale Getirme (Bell Durumu)**

Dolanıklık oluşturmak için **Hadamard (H)** kapısı ve **Kontrollü-NOT (CNOT)** kapısı kullanılır.

```
import cirq

# 1. İki qubit tanımlayalım
qubit_a = cirq.GridQubit(0, 0)
qubit_b = cirq.GridQubit(0, 1)

# 2. Kuantum Devresini oluşturalım
circuit = cirq.Circuit(
    # a) Qubit A'yı süperpozisyona getir (H)
    cirq.H(qubit_a),
    # b) CNOT: Qubit A kontrol, Qubit B hedef.
    # Eğer A = 1 ise, B'nin durumunu çevir (NOT)
    cirq.CNOT(qubit_a, qubit_b),
    # Ölçüm
    cirq.measure(qubit_a, key='a'),
    cirq.measure(qubit_b, key='b')
)

# 3. Devreyi simüle edelim (1000 kez)
simulator = cirq.Simulator()
results = simulator.run(circuit, repetitions=1000)

print("Kuantum Devresi:\n", circuit)
print("\nÖlçüm Sonuçları (1000 Çalıştırma):")
print(results.histogram(key='a')) # Qubit A sonuçları
print(results.histogram(key='b')) # Qubit B sonuçları
print("\nKombine Sonuçlar (a ve b):")
print(results.data.groupby(['a', 'b']).size().to_dict())
```

**Çıktı Yorumu:**
Bu işlem, Bell Durumu denen $\frac{1}{\sqrt{2}}(|00\rangle + |11\rangle)$ dolanık durumunu yaratır.
Kombine sonuçlarda sadece **(0, 0)** ve **(1, 1)** durumlarının gözlemlendiğini fark edeceksiniz. Yani **A** qubit'i **0** ise, **B** de anında **0**'dır ve tam tersi. Aralarında mükemmel bir korelasyon vardır.

## Kuantum Kapıları ve Devreler

### 1\. Kuantum Kapıları (Quantum Gates)

Kuantum kapıları, **qubit durumlarını değiştiren** temel kuantum operasyonlardır. Klasik bilgisayarlardaki **AND, OR, NOT** gibi mantık kapılarının kuantum karşılıklarıdır.

  * Kuantum kapıları, **üniter matrislerle** temsil edilir (yani tersine çevrilebilir ve bilgi kaybına neden olmazlar).

#### Sık Kullanılan Kuantum Kapıları

| Kapı Adı | İşlevi | Matematiksel Temsili (Matris) | Cirq Fonksiyonu |
| :--- | :--- | :--- | :--- |
| **NOT/X** | Qubit'in durumunu tersine çevirir ($|0\rangle \to |1\rangle$) | $X = \begin{pmatrix} 0 & 1 \\ 1 & 0 \end{pmatrix}$ | `cirq.X` |
| **Hadamard (H)** | Süperpozisyon yaratır ($|0\rangle \to \frac{|0\rangle+|1\rangle}{\sqrt{2}}$) | $H = \frac{1}{\sqrt{2}}\begin{pmatrix} 1 & 1 \\ 1 & -1 \end{pmatrix}$ | `cirq.H` |
| **PhaSe (S)** | Fazı değiştirir ($|1\rangle \to i|1\rangle$) | $S = \begin{pmatrix} 1 & 0 \\ 0 & i \end{pmatrix}$ | `cirq.S` |
| **CNOT** | Kontrollü NOT. İki qubit'i dolanık hale getirmede kullanılır. | - | `cirq.CNOT` |

### 2\. Kuantum Devresi (Quantum Circuit)

Kuantum devresi, belirli bir hesaplama yapmak için qubit'ler üzerine uygulanan, kronolojik sıradaki kuantum kapılarının bir dizisidir.

**Devre $=$ Qubit'ler $+$ Kapılar $+$ Ölçüm**

**Cirq Kod Örneği: NOT Kapısı Uygulaması**

$|0\rangle$ durumundaki qubit'e $X$ (NOT) kapısı uygulayarak onu $|1\rangle$ durumuna getirelim.

```
import cirq

# Qubit'i tanımla
q = cirq.GridQubit(0, 0)

# Devreyi oluştur: X (NOT) kapısı ve ölçüm
circuit = cirq.Circuit(
    cirq.X(q),  # Qubit'in durumunu tersine çevir: |0> -> |1>
    cirq.measure(q, key='output')
)

# Simülasyon
simulator = cirq.Simulator()
results = simulator.run(circuit, repetitions=100)

print("NOT Kapısı Devresi:\n", circuit)
print("\nÖlçüm Sonuçları (100 Çalıştırma):")
# Çıktıda sadece 1 sayısının 100 kez gözlemlenmesi beklenir.
print(results.histogram(key='output'))
```

## Simülasyon ve Algoritmalara Giriş

### 1\. Kuantum Devrelerinin Simülasyonu

Gerçek bir kuantum bilgisayara erişim zor ve pahalı olabilir. Bu yüzden, başlangıçta küçük devreleri geleneksel bilgisayarlarımızda **simüle ederiz**. Cirq, bu simülasyonu sağlayan güçlü bir araçtır.

Simülasyon, devrenin çalıştığında kuantum durumunun nasıl değiştiğini matematiksel olarak takip eder.

**Kod Örneği: Durum Vektörü Simülasyonu**

Bazen ölçüm sonuçlarından ziyade, ölçümden **önce** qubit'in kesin durumunu (olasılık genliklerini) görmek isteriz.

```
import cirq
import numpy as np

qubit = cirq.GridQubit(0, 0)

# H (Süperpozisyon) ve S (Faz) kapıları uygulayalım
circuit = cirq.Circuit(
    cirq.H(qubit),
    cirq.S(qubit)
)

# Durum Vektörü Simülatörünü kullanalım
simulator = cirq.Simulator()
# Devreyi çalıştır ve sonuç durumunu (vektör) al
final_state_vector = simulator.simulate(circuit).final_state_vector

print("Devre:\n", circuit)
print("\nSon Durum Vektörü (Ölçüm Öncesi):")
print(final_state_vector)

# Sonucu Yorumlama:
# final_state_vector = [ |0> genliği, |1> genliği ]
# Bu durumda: [ 0.707 + 0.j , 0. + 0.707j ]
# Durum: $\frac{1}{\sqrt{2}}|0\rangle + i \frac{1}{\sqrt{2}}|1\rangle$
# Ölçüm olasılıkları: $|0\rangle$ için $(1/\sqrt{2})^2 = 0.5$, $|1\rangle$ için $|i/\sqrt{2}|^2 = 0.5$
```

### 2\. Geleneksel Kuantum Algoritmalarına Giriş

Kuantum hesaplamanın asıl gücü, geleneksel algoritmaların çok uzun sürede çözdüğü sorunları, kuantum prensiplerini kullanarak katlanarak daha hızlı çözebilen algoritmalar geliştirmesinden gelir.

#### a) Grover Algoritması (Arama)

  * **Ne yapar:** Sıralanmamış bir veritabanında belirli bir öğeyi daha hızlı bulur.
  * **Hızlanma:** Klasik olarak $N$ öğe için $\mathcal{O}(N)$ adım gerekirken, Grover $\mathcal{O}(\sqrt{N})$ adımda bulabilir. Bu, kuadratik bir hızlanmadır.

#### b) Shor Algoritması (Çarpanlara Ayırma)

  * **Ne yapar:** Çok büyük sayıları asal çarpanlarına ayırır.
  * **Önemi:** Günümüzdeki internet güvenliğinin temelini oluşturan **RSA şifreleme** sistemi, büyük sayıları çarpanlara ayırmanın klasik bilgisayarlar için çok uzun sürmesi prensibine dayanır. Shor algoritması, kuantum bilgisayarlar için bu süreyi kısaltarak RSA'yı kırabilir.

#### c) Kuantum Fourier Dönüşümü (QFT)

  * Shor algoritması gibi birçok kuantum algoritmasının temel yapı taşıdır. Klasik Fourier Dönüşümünün kuantum versiyonudur ve kuantum durumlarının faz bilgisi üzerinde çalışır.
