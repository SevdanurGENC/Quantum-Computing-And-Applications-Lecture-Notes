
## Qiskit ile Kuantum Algoritması Programlamaya Giriş

Qiskit, kuantum devreleri tasarlamak ve bunları simüle etmek veya gerçek kuantum donanımlarında çalıştırmak için kullanılan bir Python çerçevesidir.

### Temel Qiskit Akışı

Qiskit'teki her kuantum programı genellikle şu 4 adımı izler:

1.  **Map (Haritalama):** Qubit'leri ve klasik bit'leri tanımla.
2.  **Circuit (Devre Kurma):** Qubit'lere kuantum kapılarını uygula.
3.  **Execute (Çalıştırma):** Devreyi bir simülatörde veya gerçek bir kuantum bilgisayarda çalıştır.
4.  **Analyze (Analiz):** Ölçüm sonuçlarını analiz et ve görselleştir.

```
# Gerekli modülleri içe aktaralım
from qiskit import QuantumCircuit, transpile
from qiskit_aer import Aer # Import Aer from qiskit_aer
from qiskit.visualization import plot_histogram
import matplotlib.pyplot as plt
```

### Adım 2: Süperpozisyon Devresi (Hadamard Kapısı)

Daha önce Cirq'te gördüğümüz gibi, bir qubit'i temel durumlardan ($|0\rangle$ veya $|1\rangle$) ayırıp süperpozisyona getiren en temel kapı **Hadamard (H)** kapısıdır.

#### Kod Örneği: $|+\rangle$ Durumu Yaratma

```
# 1. Map: 1 qubit ve 1 klasik bit içeren bir devre oluştur
# Klasik bit, ölçüm sonucunu saklamak içindir.
qc_superposition = QuantumCircuit(1, 1)

# 2. Circuit: 0. qubite (q[0]) Hadamard kapısı (H) uygula
# Durum |0> -> 1/sqrt(2) * (|0> + |1>) yani |+> olur.
qc_superposition.h(0)

# Devreyi çizdirelim (qiskit'in çizim aracı)
print("### Süperpozisyon Devresi ###")
print(qc_superposition)

# 3. Execute: Ölçüm yap
qc_superposition.measure(0, 0) # Qubit 0'ı Klasik Bit 0'a ölç

# 4. Analyze: Simülatörde çalıştır
simulator = Aer.get_backend('qasm_simulator') # Olasılık bazlı simülatör
compiled_circuit = transpile(qc_superposition, simulator)
job = simulator.run(compiled_circuit, shots=1024) # 1024 kez çalıştır
result = job.result()
counts = result.get_counts(qc_superposition)

print("\nÖlçüm Sonuçları (1024 Çalıştırma):", counts)

# Sonuçları görselleştir (50-50 olmasını bekliyoruz)
plot_histogram(counts).show()
# Not: Jupyter Notebook veya ortamınızın izin verdiği şekilde görselleştirme çalışacaktır.
```

### Adım 3: Temel Kuantum Algoritması: Dolanıklık (Bell Durumu)

Kuantum hesaplamanın gücünü gösteren ikinci temel yapı, iki qubit'i birbirine dolayan Bell durumudur. Bu, **Hadamard (H)** ve **Kontrollü-NOT (CX/CNOT)** kapıları ile yapılır.

#### Kod Örneği: $|\Phi^{+}\rangle$ Bell Durumu Yaratma

```
# 1. Map: 2 qubit ve 2 klasik bit içeren bir devre oluştur
qc_bell = QuantumCircuit(2, 2)

# 2. Circuit: Dolanıklığı oluştur
# a) 0. qubite H uygula (Süperpozisyon)
qc_bell.h(0)

# b) Qubit 0 (Kontrol) ve Qubit 1 (Hedef) arasında CNOT uygula
qc_bell.cx(0, 1)

# Devreyi çizdirelim
print("\n### Bell Durumu Devresi (|00> ve |11>) ###")
print(qc_bell)

# 3. Execute: Ölçüm yap
qc_bell.measure([0, 1], [0, 1]) # Qubit 0'ı Klasik Bit 0'a, Qubit 1'i Klasik Bit 1'e

# 4. Analyze: Simülatörde çalıştır
compiled_circuit = transpile(qc_bell, simulator)
job = simulator.run(compiled_circuit, shots=1024)
result = job.result()
counts = result.get_counts(qc_bell)

print("\nÖlçüm Sonuçları (1024 Çalıştırma):", counts)

# Beklenen Sonuç: Sadece '00' ve '11' durumları gözlemlenmeli.
# |01> ve |10> durumları gözlenmez, bu dolanıklık kanıtıdır.
plot_histogram(counts).show()
```

### Adım 4: Kuantum Algoritmasına Giriş (Faz Geri Beslemesi)

Kuantum algoritmalarının çoğu, kuantum kapılarının **faz** üzerindeki etkisini kullanır.

#### Kod Örneği: Z Kapısı ile Fazı Değiştirme

Bir qubit'in durumunu **Pauli-Z (Z)** kapısıyla değiştirerek, onu **Faz Bazında** inceleyebiliriz.

**İşlem:** $|+\rangle = \frac{1}{\sqrt{2}}(|0\rangle + |1\rangle)$ durumuna Z uygulayalım.
$$Z|+\rangle = \frac{1}{\sqrt{2}}(Z|0\rangle + Z|1\rangle) = \frac{1}{\sqrt{2}}(|0\rangle - |1\rangle) = |-\rangle$$

$|-\rangle$ durumuna tekrar $H$ uygulandığında sonuç **$|1\rangle$** olur. (Oysa $|+\rangle$ durumuna $H$ uygulansaydı $|0\rangle$ olurdu.)

```
# 1. Map
qc_phase = QuantumCircuit(1, 1)

# 2. Circuit: |+> durumunu oluştur
qc_phase.h(0)

# 3. Z Kapısını Uygula (Fazı değiştir)
qc_phase.z(0) # Durum |+> -> |-> olur

# 4. H Uygula (Fazı ölçülebilir genliğe çevir)
qc_phase.h(0) # Durum |-> -> |1> olur

# Devreyi çizdirelim
print("\n### Faz Geri Beslemesi (Z Kapısı) ###")
print(qc_phase)

# 5. Execute & Measure
qc_phase.measure(0, 0)
compiled_circuit = transpile(qc_phase, simulator)
job = simulator.run(compiled_circuit, shots=1024)
result = job.result()
counts = result.get_counts(qc_phase)

print("\nÖlçüm Sonuçları (1024 Çalıştırma):", counts)

# Beklenen Sonuç: Neredeyse tamamen '1' çıkması gerekir.
plot_histogram(counts).show()
```
